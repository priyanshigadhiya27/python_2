```python
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
```

```python
with open('demo.html','r',encoding='utf-8') as f:
    res=f.read()
soup=BeautifulSoup(res,'html.parser')
```

```python
box=soup.find_all('div',class_='yKfJKb row')
print(len(box))
```

```python
d={'name':[],'rom':[],'ram':[],'color':[],'stars':[],'review':[],'rear_camera':[],'front_camera':[],'old_price':[],'new_price':[],'discount':[],
'rating':[],'display(cm)':[],'display(inch)':[]}
for i in box:
    d['name'].append(i.find('div',class_='KzDlHZ').text.strip().split("(")[0].strip())
    d['color'].append(i.find('div',class_='KzDlHZ').text.strip().split("(")[1].split(",")[0])
    d['stars'].append(i.find('div',class_='XQDdHH').text.strip().split("(")[0].strip())
    d['rating'].append(i.find('span',class_='Wphh3N').text.strip().split("&")[0].split()[0].replace(",",''))
    d['review'].append(i.find('span',class_='Wphh3N').text.strip().split("&")[1].split()[0])
    d['ram'].append(i.find('ul',class_="G4BRas").text.split('|')[0].strip())
    d['rom'].append(i.find('ul',class_="G4BRas").text.split('|')[1].split('ROM')[0].strip())
    d['display(cm)'].append(i.find('ul',class_="G4BRas").text.split('|')[1].split()[2].replace('ROM',''))
    d['display(inch)'].append(i.find('ul',class_="G4BRas").text.split('|')[1].split()[4].replace('(',''))
    d['rear_camera'].append(i.find('ul',class_="G4BRas").text.split('|')[1].split("Display")[-1])
    d['front_camera'].append(i.find('ul',class_="G4BRas").text.split('|')[2].split()[0])
    d['old_price'].append(i.find('div',class_='Nx9bqj _4b5DiR').text.strip()[1:].replace(",",''))
    try:
        d['new_price'].append(i.find('div',class_="yRaY8j ZYYwLA").text.strip()[1:].replace(",",''))
    except:
        d['new_price'].append(np.nan)

    try:
        d['discount'].append(i.find('div',class_='UkUFwK').text.split("off")[0])
    except:
        d['discount'].append(np.nan)
        
df=pd.DataFrame(d)
df
```
